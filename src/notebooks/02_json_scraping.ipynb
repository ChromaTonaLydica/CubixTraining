```python
from bs4 import BeautifulSoup
import pandas as pd
import requests

url="https://en.wikipedia.org/wiki/Community_areas_in_Chicago"
headers = {"User-Agent": "Mozilla/5.0 (compatible; MyScraper/1.0)"}

response = requests.get(url=url, headers=headers)
soup = BeautifulSoup(response.content, "html.parser")

response 
```

```python
print(soup.prettify())
```

```python
soup.find_all("table")
```

```python
for table in soup.find_all("table"):
    print(table.get("class"))
```

```python
table = soup.find("table", class_="wikitable sortable plainrowheaders mw-datatable")
```

```python
table
```

```python
for row in table.tbody.find_all("tr")[2:-1]:
    print(row)
    print("-" * 50)
```

```python
for row in table.tbody.find_all("tr")[2:-1]:
    cells = row.find_all("td")
    print(cells[0].get_text(strip=True))
```

```python
for row in table.tbody.find_all("tr")[2:-1]:
    header_cells = row.find_all("th")
    print(header_cells[0].get_text(strip=True))
```

```python
for row in table.tbody.find_all("tr")[2:-1]:
    cells = row.find_all("td")
    header_cells = row.find_all("th")

    print(cells[0].get_text(strip=True))
    print(header_cells[0].get_text(strip=True))
    print()
```

```python
data = []

for row in table.tbody.find_all("tr")[2:-1]:
    cells = row.find_all("td")
    header_cells = row.find_all("th")

    area_code = cells[0].get_text(strip=True)
    community_area_name = header_cells[0].get_text(strip=True)

    data.append({"area_code": area_code, "community_area": community_area_name})

data
```

```python
community_areas_df = pd.DataFrame(data)
```

```python
community_areas_df
```

```python
community_areas_df["area_code"] = community_areas_df["area_code"].astype("int")
```
