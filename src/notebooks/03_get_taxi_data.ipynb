# endpoint expects a SQL query inside the query parameter.

It does not use the old Socrata parameters.
Use query= and put the SQL inside it.
Spaces and special characters in SQL should be URL encoded.

```python
url = "https://data.cityofchicago.org/api/v3/views/ajtu-isnz/query.json"

params = {
    "app_token": "KtrnoD09nF0tQuc1PWsTC2vgo",
    "query": """
        SELECT *
        WHERE `trip_start_timestamp` >= '2024-02-16T00:00:00'
        AND `trip_start_timestamp` <= '2024-02-16T23:59:59'
        LIMIT 30000
    """
}

response = requests.get(url, params=params)
data = response.json()

data

```

#### Getting a specific day's data

```python
current_datetime = datetime.now() - relativedelta(months=2)
formatted_datetime = current_datetime.strftime("%Y-%m-%d")

formatted_datetime
```

```python
url = "https://data.cityofchicago.org/api/v3/views/ajtu-isnz/query.json"

params = {
    "app_token": "KtrnoD09nF0tQuc1PWsTC2vgo",
    "query": f"""
        SELECT *
        WHERE `trip_start_timestamp` >= '{formatted_datetime}T00:00:00'
        AND `trip_start_timestamp` <= '{formatted_datetime}T23:59:59'
        LIMIT 30000
    """
}

print(params["query"])

response = requests.get(url, params=params)
data = response.json()

data
```
